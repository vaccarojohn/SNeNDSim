```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read NumPy data
infile_dir = 'graph_data'
signal_segment_data = np.load('../signal/' + infile_dir + '/signal_segment_data.npz')
signal_selection_data = np.load('../../edepsim/signal/' + infile_dir + '/signal_selection_data.npz')

cosmics_segment_data = np.load('../cosmics/' + infile_dir + '/cosmics_segment_data.npz')
cosmics_selection_data = np.load('../../edepsim/cosmics/' + infile_dir + '/cosmics_selection_data.npz')

BRN_segment_data = np.load('../BRN/' + infile_dir + '/BRN_segment_data.npz')
BRN_selection_data = np.load('../../edepsim/BRN/' + infile_dir + '/BRN_selection_data.npz')

outside_segment_data = np.load('../outside/' + infile_dir + '/outside_segment_data.npz')
outside_selection_data = np.load('../../edepsim/outside/' + infile_dir + '/outside_selection_data.npz')

print("Successfully loaded data!")
```

```python
# Get number of signal, cosmic, and BRN events in the active volume

print("Cut #1")
print("Number of Signal Events in TPC: " + str(len(signal_segment_data['psenergy'])))
#print("Number of Cosmic Events in TPC: " + str(len(cosmics_segment_data['psenergy'])))
print("Number of BRN Events in TPC: " + str(len(BRN_segment_data['psenergy'])))
print("Number of outside Events in TPC: " + str(len(outside_segment_data['psenergy'])))
```

```python
# Apply selection cut [SignalVolEFrac > 0.33 & MaxE > 10.1 & TotalE < 52 & MaxELength > 2.5 & CRTTotal < 6 & tMin >= 0 & tMin <= 8.1]

n = 0
for psenergy, lcte, aenergy, lctl, crt, tmin, closestedge in np.column_stack((signal_segment_data['psenergy'], signal_segment_data['lctes'], signal_segment_data['aenergy'], signal_segment_data['lctls'], signal_selection_data['crt'], signal_selection_data['tmin'], signal_segment_data['closestedge'])):
    if psenergy > 0.33 and lcte > 10.1 and aenergy < 52 and lctl > 4.5 and crt < 6 and tmin >= 0 and tmin <= 8.1 and closestedge > 2.5:
        n += 1

print("Number of Signal Events with [SignalVolEFrac > 0.33 & MaxE > 10.1 & TotalE < 52 & MaxELength > 2.5 & CRTTotal < 6 & tMin >= 0 & tMin <= 8.1]: " + str(n))

"""n = 0
for psenergy, lcte, aenergy, lctl, crt, tmin, closestedge in np.column_stack((cosmics_segment_data['psenergy'], cosmics_segment_data['lctes'], cosmics_segment_data['aenergy'], cosmics_segment_data['lctls'], cosmics_selection_data['crt'], cosmics_selection_data['tmin'], cosmics_segment_data['closestedge'])):
    if psenergy > 0.4:
        n += 1

print("Number of Cosmic Events with [SignalVolEFrac > 0.33 & MaxE > 10.1 & TotalE < 52 & MaxELength > 2.5 & CRTTotal < 6 & tMin >= 0 & tMin <= 8.1]: " + str(n))"""

n = 0
for psenergy, lcte, aenergy, lctl, crt, tmin, closestedge in np.column_stack((BRN_segment_data['psenergy'], BRN_segment_data['lctes'], BRN_segment_data['aenergy'], BRN_segment_data['lctls'], BRN_selection_data['crt'], BRN_selection_data['tmin'], BRN_segment_data['closestedge'])):
    if psenergy > 0.33 and lcte > 10.1 and aenergy < 52 and lctl > 4.5 and crt < 6 and tmin >= 0 and tmin <= 8.1 and closestedge > 2.5:
        n += 1

print("Number of BRN Events with [SignalVolEFrac > 0.33 & MaxE > 10.1 & TotalE < 52 & MaxELength > 2.5 & CRTTotal < 6 & tMin >= 0 & tMin <= 8.1]: " + str(n))

n = 0
for psenergy, lcte, aenergy, lctl, crt, tmin, closestedge in np.column_stack((outside_segment_data['psenergy'], outside_segment_data['lctes'], outside_segment_data['aenergy'], outside_segment_data['lctls'], outside_selection_data['crt'], outside_selection_data['tmin'], outside_segment_data['closestedge'])):
    if psenergy > 0.33 and lcte > 10.1 and aenergy < 52 and lctl > 4.5 and crt < 6 and tmin >= 0 and tmin <= 8.1 and closestedge > 2.5:
        n += 1

print("Number of outside Events with [SignalVolEFrac > 0.33 & MaxE > 10.1 & TotalE < 52 & MaxELength > 2.5 & CRTTotal < 6 & tMin >= 0 & tMin <= 8.1]: " + str(n))
```
