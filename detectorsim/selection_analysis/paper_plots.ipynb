```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read NumPy data
infile_dir = 'graph_data'
signal_segment_data = np.load('../signal/' + infile_dir + '/signal_segment_data.npz')
signal_selection_data = np.load('../../edepsim/signal/' + infile_dir + '/signal_selection_data.npz')

cosmics_segment_data = np.load('../cosmics/' + infile_dir + '/cosmics_segment_data.npz')
cosmics_selection_data = np.load('../../edepsim/cosmics/' + infile_dir + '/cosmics_selection_data.npz')

BRN_segment_data = np.load('../BRN/' + infile_dir + '/BRN_segment_data.npz')
BRN_selection_data = np.load('../../edepsim/BRN/' + infile_dir + '/BRN_selection_data.npz')

outside_segment_data = np.load('../outside/' + infile_dir + '/outside_segment_data.npz')
outside_selection_data = np.load('../../edepsim/outside/' + infile_dir + '/outside_selection_data.npz')

print("Successfully loaded data!")
```

```python
weights_signal = np.ones_like(signal_segment_data['aenergy']) / len(signal_segment_data['aenergy'])
weights_cosmics = np.ones_like(cosmics_segment_data['aenergy']) / len(cosmics_segment_data['aenergy'])
weights_BRN = np.ones_like(BRN_segment_data['aenergy']) / len(BRN_segment_data['aenergy'])
weights_outside = np.ones_like(outside_segment_data['aenergy']) / len(outside_segment_data['aenergy'])
```

```python
# Total energy deposited per event (in active volume)

# Define bins
bins = np.arange(0, 101, 1)
plt.hist(signal_segment_data['aenergy'], bins=bins, histtype='step', edgecolor='blue', weights=weights_signal, label='Signal')
plt.hist(cosmics_segment_data['aenergy'], bins=bins, histtype='step', edgecolor='orange', weights=weights_cosmics, label='Cosmics')
plt.hist(BRN_segment_data['aenergy'], bins=bins, histtype='step', edgecolor='black', weights=weights_BRN, label='BRN')
plt.hist(outside_segment_data['aenergy'], bins=bins, histtype='step', edgecolor='purple', weights=weights_outside, label='CC Outside Fiducial Volume')
plt.yscale('log')
plt.xlim(0, 100)

# Axis labels
plt.xlabel('Energy Deposited (MeV)')
plt.ylabel('Counts (Unit Normalized)')
plt.title('Observed Energy in TPC')
plt.legend()

plt.show()
```
