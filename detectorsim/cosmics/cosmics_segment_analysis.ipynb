```python
import h5py
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Read NumPy data
infile_dir = 'graph_data'
data = np.load(infile_dir + '/cosmics_segment_data.npz')
data2 = np.load('../../edepsim/cosmics/' + infile_dir + '/cosmics_segment_data_exp.npz')

print("Successfully loaded data!")
```

```python
# Total energy deposited per event (in active volume)

# Define bins
bins = np.arange(0, 2020, 20)
plt.hist(data['aenergy'], bins=bins, edgecolor='black')
plt.xlim(0, 2000)

# Axis labels
plt.yscale('log')
plt.xlabel('Energy Deposited (MeV)')
plt.ylabel('Frequency')
plt.title('Total Energy Deposited per Event (in Active Volume)')

plt.show()
```

```python
# Percent difference between reconstructed and true energy (in active volume)

# Define bins
bins = np.arange(-1, 1.025, 0.025)
plt.hist(data['aenergy'] / data2['aenergy'] - 1, bins=bins, edgecolor='black')
plt.xlim(-1, 1)

# Axis labels
plt.xlabel('Percent Difference')
plt.ylabel('Frequency')
plt.title('Difference between Reconstructed and True Energy in Active Volume per Event')

plt.show()
```
